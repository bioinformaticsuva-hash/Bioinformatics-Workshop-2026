## 1. Raw vs. Filtered Feature-Barcode Matrices

Cell Ranger's `outs/` folder contains **two versions** of the count matrix, and the difference between them matters a lot for QC:

- **`raw_feature_bc_matrix/`**
  - Contains **every barcode** Cell Ranger observed — typically several hundred thousand to a few million barcodes per sample.
  - The overwhelming majority of these are **empty droplets** (no real cell, just background/ambient mRNA + noise).
  - Same three files as filtered: `barcodes.tsv.gz`, `features.tsv.gz`, `matrix.mtx.gz`.
  - Useful for: estimating background/ambient signal, re-running custom cell calling, and — relevant here — **SoupX correction**.

- **`filtered_feature_bc_matrix/`**
  - Contains only the barcodes Cell Ranger's cell-calling algorithm decided are **real cells** (the "knee" in the barcode-rank plot).
  - This is the matrix most people load directly into Seurat for standard analysis.
  - Same three file types, just far fewer columns (barcodes).

- **Why keep both around?** The filtered matrix alone can't tell you what the *background* looks like — you need the raw matrix (specifically, the empty droplets in it) to estimate that. That's exactly what SoupX uses it for.

## 2. Why Ambient RNA Correction Is Needed

- During tissue dissociation, some cells lyse or become stressed and **release free-floating mRNA** into the surrounding solution.
- In droplet-based scRNA-seq, that free-floating "soup" of mRNA gets captured **in every droplet** — including droplets that contain a real cell.
- Result: every cell's expression profile is a mix of **its own real transcripts** + **a low background of transcripts from other cells in the sample** (often highly-expressed, cell-type-specific genes like hemoglobin from lysed red blood cells, or immunoglobulins from plasma cells).
- This contamination is usually small per gene, but because it's present in *every* cell, it can:
  - Make cells look like they weakly express marker genes of a completely different cell type
  - Blur cluster boundaries and reduce marker-gene specificity
  - Distort differential expression results between conditions (e.g., pre- vs. post-treatment) if ambient composition differs between samples

This is exactly the kind of problem that's invisible in bulk RNA-seq and only becomes obvious — and fixable — at single-cell resolution.


## 3. What SoupX Does

**SoupX** (Young & Behjati, *GigaScience* 2020) estimates and removes the ambient RNA ("soup") contribution from each cell's counts, **before** normalization/clustering.

At a high level, SoupX:

1. **Estimates the soup profile** — the background expression pattern — using the empty droplets in the **raw matrix** (droplets present in `raw_feature_bc_matrix` but not called as cells in `filtered_feature_bc_matrix`). These empty droplets contain (almost) nothing but ambient RNA, so their average expression profile is a good proxy for "what the soup looks like."
2. **Estimates the contamination fraction (ρ, rho)** for each cell — roughly, what fraction of that cell's UMIs are ambient rather than real. SoupX does this by looking at genes that should be **mutually exclusive** across cell types (e.g., hemoglobin genes should be ~absent in T cells) — any expression of those genes in a cell that clearly isn't that cell type is attributed to ambient contamination. `autoEstCont()` automates finding good genes for this and estimating rho per cell.
3. **Subtracts the estimated ambient contribution** from each cell's counts proportionally to that cell's estimated rho and the soup profile, then rounds back to integer counts. This produces a **corrected count matrix**.
4. The corrected matrix is what gets used to build (or replace the counts in) the Seurat object — **not** the raw filtered counts.

### Why this needs *both* matrices
- Filtered matrix → which barcodes are cells, and their observed (contaminated) counts
- Raw matrix → what the background/ambient signal looks like (from the empty droplets)
- SoupX needs both simultaneously to separate "real signal" from "ambient contribution" per cell.


````r
# Make a Seurat object from the sparce matrix as follows:
srat  <- CreateSeuratObject(counts = filtd.matrix)
srat
````

````r
filt_genes <- rownames(filtd.matrix)
````

````r
# Subset raw.matrix to keep only the genes in filt.matrix
# Since the provided table of droplets (tod) and table of counts (toc) have different numbers of genes.  Both tod and toc must have the same genes in the same order.
raw.matrix.subset <- raw.matrix[rownames(raw.matrix) %in% filt_genes, ]
````

````r
# Let’s make a “SoupChannel”, the object needed to run SoupX.
preT_data_SoupX <- SoupChannel(raw.matrix.subset, filtd.matrix, calcSoupProfile = TRUE)
preT_data_SoupX
````

````r
srat <- SCTransform(srat, verbose = F)
srat <- RunPCA(srat, verbose = F)
srat <- RunUMAP(srat, dims = 1:30, verbose = F)
srat <- FindNeighbors(srat, dims = 1:30, verbose = F)
srat <- FindClusters(srat, verbose = T)
````

````r
DefaultAssay(srat) <- "SCT"
````

````r
# After clustering is obtained, it can be added to the channel using setClusters. setDR is useful for visualizations.
meta    <- srat@meta.data
umap    <- srat@reductions$umap@cell.embeddings
preT_data_SoupX  <- setClusters(preT_data_SoupX, setNames(meta$seurat_clusters, rownames(meta)))
preT_data_SoupX  <- setDR(preT_data_SoupX, umap)
#head(meta)
````

````r
options(repr.plot.width = 20, repr.plot.height = 15, repr.plot.res = 100)
#preT_data_SoupX = estimateSoup(preT_data_SoupX)
preT_data_SoupX = autoEstCont(preT_data_SoupX)
````

## Correcting expression profile
We have now calculated or set the contamination fraction for each cell and would like to use this to remove the contamination from the original count matrix. As with estimating the contamination, this procedure is made much more robust by providing clustering information. This is because there is much more power to separate true expression from contaminating expression when counts are aggregated into clusters. Furthermore, the process of redistributing corrected counts from the cluster level to individual cells automatically corrects for variation in the cell specific contamination rate.
We have already loaded clustering information into our sc object with setClusters, which will be used by default. So we can just run.

**Note:** `autoEstCont()` internally runs a quick clustering pass to help estimate rho; you can also supply your own clusters (`setClusters()`) if you already have some from a prior analysis pass.

````r
preT_data_SoupX_out = adjustCounts(preT_data_SoupX)
````

## Investigating changes in expression
Before proceeding let’s have a look at what this has done. We can get a sense for what has been the most strongly decreased by looking at the fraction of cells that were non-zero and now set to zero after correction.

````r
cntSoggy = rowSums(preT_data_SoupX$toc > 0)
cntStrained = rowSums(preT_data_SoupX_out > 0)
mostZeroed = tail(sort((cntSoggy - cntStrained)/cntSoggy), n = 10)
mostZeroed
````

## 4. Where This Fits in the Pipeline

```
raw_feature_bc_matrix/      ──┐
                              ├──▶  SoupX (estimate soup + rho, adjustCounts()) ──▶ corrected counts ──▶ CreateSeuratObject() ──▶ standard QC/normalization/clustering
filtered_feature_bc_matrix/ ──┘
```

Ambient correction happens **early** — right after loading the Cell Ranger output and before normalization, clustering, or marker identification — since contamination affects the input to every step downstream of it.

````r
obj_combined.integrated <- FindNeighbors(obj_combined.integrated, reduction = "pca", dims = 1:50)
dims    <- 1:5
res_vec <- round(seq(0.1, 1.0, by = 0.1), 1)

emb <- Embeddings(obj_combined.integrated[["pca"]])[, dims]
set.seed(42)
idx <- if (nrow(emb) > 8000) sort(sample(nrow(emb), 8000)) else seq_len(nrow(emb))
d   <- dist(emb[idx, ])

sil_df <- bind_rows(lapply(res_vec, function(r) {
  message("resolution ", r)
  tmp <- FindClusters(obj_combined.integrated, resolution = r, verbose = FALSE)
  cl  <- as.numeric(tmp$seurat_clusters)[idx]
  rm(tmp); gc()
  if (length(unique(cl)) < 2)
    return(data.frame(resolution = r, mean_sil = NA_real_, n_clusters = 1L))
  data.frame(resolution  = r,
             mean_sil    = mean(silhouette(cl, dist = d)[, 3]),
             n_clusters  = length(unique(cl)))
}))

sil_df
````

## Finding Neighbors and Clusters - Choosing a clustering resolution: the within-cluster sum of squares (WCSS) elbow

**What we compute.** For each resolution from 0.1 to 1.0 we cluster the cells, then
measure how tightly packed each cluster is in PCA space. For every cluster we find its
centroid and sum the squared distances from each of its cells to that centroid; adding
this across all clusters gives the within-cluster sum of squares. Low WCSS means cells
sit close to their own cluster's centre — a compact partition.

````r

res_vec <- round(seq(0.1, 1.0, by = 0.1), 1)
X <- Embeddings(obj_combined.integrated[["pca"]])[, 1:5]

wss_for <- function(cl, X) {
  sum(sapply(split(seq_along(cl), cl), function(i) {
    if (length(i) < 2) return(0)
    m <- colMeans(X[i, , drop = FALSE])
    sum(sweep(X[i, , drop = FALSE], 2, m)^2)
  }))
}

elbow_df <- bind_rows(lapply(res_vec, function(r) {
  message("resolution ", r)
  tmp <- FindClusters(obj_combined.integrated, resolution = r, verbose = FALSE)
  cl  <- as.character(tmp$seurat_clusters)
  rm(tmp); gc()
  data.frame(resolution = r, n_clusters = length(unique(cl)), wss = wss_for(cl, X))
})) %>%
  mutate(pct_drop = 100 * (lag(wss) - wss) / lag(wss))

elbow_df

ggplot(elbow_df, aes(resolution, wss)) +
  geom_line(linewidth = 0.8, colour = 'grey40') +
  geom_point(size = 3) +
  geom_text(aes(label = n_clusters), vjust = -1.2, size = 4.5) +
  scale_x_continuous(breaks = res_vec) +
  labs(x = 'Resolution', y = 'Within-cluster sum of squares',
       caption = 'labels = number of clusters') +
  theme_bw() +
  theme(text = element_text(size = 18), panel.grid.minor = element_blank())
````


## Cluster similarity
Here, we use the clusterModularity() function from the scran package as described here. The heatmap below shows the difference between the observed and expected edge weights between each cluster pair.


````r
options(repr.plot.width = 17, repr.plot.height = 15, repr.plot.res = 100)
obj_combined.integrated_res <- JoinLayers(obj_combined.integrated_res, assay = "RNA")
sce <- as.SingleCellExperiment(obj_combined.integrated_res, assay = "RNA")


reducedDim(sce, 'PCA_sub') <- reducedDim(sce, 'PCA')[,1:15, drop = FALSE]

g <- scran::buildSNNGraph(sce, use.dimred = 'PCA_sub')

ratio <- bluster::pairwiseModularity(g, obj_combined.integrated_res@meta.data$seurat_clusters, as.ratio = TRUE)

ratio_to_plot <- log10(ratio+1)

p <- ratio_to_plot %>%
  as_tibble() %>%
  rownames_to_column(var = 'cluster_1') %>%
  pivot_longer(
    cols = 2:ncol(.),
    names_to = 'cluster_2',
    values_to = 'probability'
  ) %>%
  mutate(
    cluster_1 = as.character(as.numeric(cluster_1) - 1),
    cluster_1 = factor(cluster_1, levels = rev(unique(cluster_1))),
    cluster_2 = factor(cluster_2, levels = unique(cluster_2))
  ) %>%
  ggplot(aes(cluster_2, cluster_1, fill = probability)) +
  geom_tile(color = 'white') +
  geom_text(aes(label = round(probability, digits = 2)), size = 6) +
  scale_x_discrete(name = 'Cluster', position = 'top') +
  scale_y_discrete(name = 'Cluster') +
  scale_fill_gradient(
    name = 'log10(ratio)', low = 'white', high = '#c0392b', na.value = '#bdc3c7',
    guide = guide_colorbar(
      frame.colour = 'black', ticks.colour = 'black', title.position = 'left',
      title.theme = element_text(hjust = 1, angle = 90),
      barwidth = 0.75, barheight = 10
    )
  ) +
  coord_fixed() +
  theme_bw() +
  theme(
    legend.position = 'right',
    panel.grid.major = element_blank()
  )
p  & theme(plot.title = element_text(size = 40, face = "bold")) & theme(text=element_text(size=25))

````
